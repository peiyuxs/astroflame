For Cloudflare R2, install `misc/requirements.txt`, copy `misc/.env.example` to `misc/.env`, and fill in the account ID and R2 access keys locally. The script streams files to R2 in multipart chunks without staging them on your computer. The `.env` file is git-ignored; never commit or share its secrets.

Run the preview cell to enter the PSI ID and inspect branch sizes. The download cell prompts separately for each branch. Local downloads are size-checked and skipped on reruns when already complete; R2 uploads are also skipped when the stored file metadata matches.

You can also run `python misc/downloadfiles.py --psi-id PSI-26 --list-only` from the repository root to preview, or omit `--list-only` to choose branches and download.
